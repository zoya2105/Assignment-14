```python
avg_order = df.groupby('city')['order_value'].mean()
plt.bar(avg_order.index, avg_order.values)
plt.xlabel('City')
plt.ylabel('Average Order Value')
plt.title('Average Order Value by City')
plt.xticks(rotation = 45)
plt.show()
```
